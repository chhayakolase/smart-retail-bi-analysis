```python
import pandas as pd
df = pd.read_csv(r"C:\Users\chhay\OneDrive\Desktop\Smart-Retail-BI-system\Dataset\SampleSuperstore.csv")
df = df.drop_duplicates()
```

```python
import matplotlib.pyplot as plt
df.groupby("Category")["Sales"].sum().plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
import pandas as pd
df = pd.read_csv(r"C:\Users\chhay\OneDrive\Desktop\Smart-Retail-BI-system\Dataset\SampleSuperstore.csv")
df = df.drop_duplicates()
df.groupby("Sub-Category")["Sales"].sum().sort_values().plot(kind="barh",figsize=(10,6))
plt.title("Sales by Sub-Category")
plt.xlabel("Sales")
plt.ylabel("Sales-Category")
plt.show()
```
